```python
import os
U="https://huggingface.co/spaces/TuwaiqAcademy/AISA-ArabicFC-SharedTask-Leaderboard/raw/main"
for f in ["normalize.py","eval_lib.py","data_loader.py"]:
    if not os.path.exists(f): os.system(f"wget -q {U}/{f}")     # download official scorer
from data_loader import parse_predictions, load_gold            # data_loader needs `datasets` (Colab has it)
from eval_lib import evaluate
gold = load_gold("dev")
for f in ["qwen_final_submission_trackA.jsonl","qwen_final_submission_trackB.jsonl"]:
    preds,_ = parse_predictions(f)
    r = evaluate(preds, gold)
    print(f"{f}: OverallA {r['overall_a']:.4f}  ArgEM {r['argem']:.4f}  OverallB {r['overall_b']:.4f}")
```
